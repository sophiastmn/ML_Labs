```python
# Vanishing Gradient Problem Demonstration - FIXED VERSION
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import time
import matplotlib.pyplot as plot;

# Set random seeds for reproducibility
np.random.seed(42)
tf.random.set_seed(42)

# ============================================
# 1. CREATE A DEEP NETWORK AND MONITOR GRADIENTS
# ============================================

def create_deep_network(activation='sigmoid', n_layers=10, neurons_per_layer=100, input_shape=10):
    """
    Create a deep neural network with specified activation function
    """
    model = keras.Sequential()
    model.add(layers.Input(shape=(input_shape,)))  # Explicit Input layer

    # Add hidden layers
    for i in range(n_layers):
        model.add(layers.Dense(neurons_per_layer,
                               activation=activation,
                               name=f'hidden_{i+1}',
                               kernel_initializer='glorot_uniform'))

    # Output layer for binary classification
    model.add(layers.Dense(1, activation='sigmoid', name='output'))

    return model

# ============================================
# 2. CUSTOM CALLBACK TO TRACK GRADIENTS - FIXED VERSION
# ============================================

class GradientMonitor(keras.callbacks.Callback):
    """Custom callback to track gradient magnitudes during training"""

    def __init__(self, log_every=5, sample_batch=None):
        super().__init__()
        self.log_every = log_every
        self.gradient_history = {}
        self.epoch_history = []
        self.sample_batch = sample_batch  # Store a sample batch for gradient computation

    def on_train_begin(self, logs=None):
        """Initialize gradient history structure"""
        for layer in self.model.layers:
            if 'hidden' in layer.name:  # Only track hidden layers
                self.gradient_history[layer.name] = []

    def on_epoch_end(self, epoch, logs=None):
        if epoch % self.log_every == 0:
            self.epoch_history.append(epoch)

            # Use provided sample batch or create a random one
            if self.sample_batch is not None:
                batch = self.sample_batch
            else:
                # Create a random batch with the correct input shape
                input_shape = self.model.input_shape[1:]
                batch = tf.random.normal((32, *input_shape))

            # Compute gradients
            with tf.GradientTape() as tape:
                # Forward pass
                predictions = self.model(batch, training=True)
                # Dummy loss (binary crossentropy with random targets)
                dummy_targets = tf.random.uniform((32, 1), maxval=2, dtype=tf.int32)
                dummy_targets = tf.cast(dummy_targets, tf.float32)
                loss = tf.keras.losses.binary_crossentropy(dummy_targets, predictions)

            # Compute gradients
            grads = tape.gradient(loss, self.model.trainable_variables)

            # Calculate gradient norms for hidden layers
            grad_index = 0
            for layer in self.model.layers:
                if 'hidden' in layer.name:
                    # Each hidden layer has weights and bias
                    if grads[grad_index] is not None:
                        grad_norm = tf.norm(grads[grad_index]).numpy()
                        self.gradient_history[layer.name].append(grad_norm)
                    grad_index += 2  # Skip bias gradient
                elif 'output' not in layer.name:
                    # Count parameters for non-hidden layers
                    grad_index += len(layer.trainable_variables)

# ============================================
# 3. GENERATE SYNTHETIC DATA
# ============================================

def generate_data(n_samples=2000, n_features=10):
    """Generate synthetic dataset"""
    X = np.random.randn(n_samples, n_features)

    # Create a non-linear relationship
    y = (np.sin(X[:, 0]) + 0.5 * X[:, 1] ** 2 +
         0.3 * X[:, 2] * X[:, 3] + 0.1 * np.random.randn(n_samples))

    # Convert to binary classification problem
    y_binary = (y > np.median(y)).astype(int)

    return X.astype(np.float32), y_binary.astype(np.float32)

def printGradientGraphs(gradientData):
  #prints out the graphs based on the gradient
  #graph the gradient data
  for i in range(len(gradientData)):
    plot.scatter([g[i] for g in gradientData], [*range(0, len(gradientData)*2, 2)]);
    plot.xlabel(f"Gradient of Error in layer {i+1}");
    plot.ylabel("Error");
    plot.show();

# Generate data
X, y = generate_data(2000, 10)
X_train, X_test = X[:1500], X[1500:]
y_train, y_test = y[:1500], y[1500:]

# Create a sample batch for gradient monitoring
sample_batch = tf.constant(X_train[:32])

# ============================================
# 4. EXPERIMENT 1: SIGMOID ACTIVATION (PRONE TO VANISHING GRADIENTS)
# ============================================

print("=" * 70)
print("EXPERIMENT 1: Deep Network with Sigmoid Activation")
print("=" * 70)

# Create deep network with sigmoid
sigmoid_model = create_deep_network(activation='sigmoid', n_layers=10, input_shape=10)

# Compile the model
sigmoid_model.compile(
    optimizer=keras.optimizers.SGD(learning_rate=0.01),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Show model summary
print("\nModel Architecture:")
sigmoid_model.summary()

# Create gradient monitor with sample batch
grad_monitor_sigmoid = GradientMonitor(log_every=2, sample_batch=sample_batch)

# Train the model
print("\nTraining deep sigmoid network...")
start_time = time.time()
history_sigmoid = sigmoid_model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=50,
    batch_size=32,
    callbacks=[grad_monitor_sigmoid],
    verbose=1  # Set to 1 to see progress
)
sigmoid_time = time.time() - start_time

print(f"\nTraining completed in {sigmoid_time:.2f} seconds")
print(f"Final training accuracy: {history_sigmoid.history['accuracy'][-1]:.4f}")
print(f"Final validation accuracy: {history_sigmoid.history['val_accuracy'][-1]:.4f}")

#get the gradient history
#fix the gradient history into a way that I can manipulate
gradientsOverTime = [[float(val) for val in grad_monitor_sigmoid.gradient_history[key]] for key in grad_monitor_sigmoid.gradient_history];
printGradientGraphs(gradientsOverTime);


# ============================================
# 5. EXPERIMENT 2: RELU ACTIVATION (MITIGATES VANISHING GRADIENTS)
# ============================================

print("\n" + "=" * 70)
print("EXPERIMENT 2: Deep Network with ReLU Activation")
print("=" * 70)

# Create deep network with ReLU
relu_model = create_deep_network(activation='relu', n_layers=10, input_shape=10)

# Compile the model
relu_model.compile(
    optimizer=keras.optimizers.SGD(learning_rate=0.01),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Create gradient monitor
grad_monitor_relu = GradientMonitor(log_every=2, sample_batch=sample_batch)

# Train the model
print("\nTraining deep ReLU network...")
start_time = time.time()
history_relu = relu_model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=50,
    batch_size=32,
    callbacks=[grad_monitor_relu],
    verbose=1
)
relu_time = time.time() - start_time

print(f"\nTraining completed in {relu_time:.2f} seconds")
print(f"Final training accuracy: {history_relu.history['accuracy'][-1]:.4f}")
print(f"Final validation accuracy: {history_relu.history['val_accuracy'][-1]:.4f}")

# ============================================
# 6. EXPERIMENT 3: EXTREMELY DEEP NETWORK (EXAGGERATED VANISHING)
# ============================================

print("\n" + "=" * 70)
print("EXPERIMENT 3: Extremely Deep Network (15 layers) with Sigmoid")
print("=" * 70)

# Create extremely deep network with sigmoid
deep_sigmoid_model = create_deep_network(activation='sigmoid', n_layers=15,
                                         neurons_per_layer=50, input_shape=10)

# Compile the model
deep_sigmoid_model.compile(
    optimizer=keras.optimizers.SGD(learning_rate=0.01),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Train briefly
print("\nTraining extremely deep sigmoid network...")
history_deep = deep_sigmoid_model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=30,
    batch_size=32,
    verbose=1
)

print(f"\nFinal training accuracy: {history_deep.history['accuracy'][-1]:.4f}")
print(f"Final validation accuracy: {history_deep.history['val_accuracy'][-1]:.4f}")

# ============================================
# 7. VISUALIZE THE VANISHING GRADIENT PROBLEM
# ============================================

plt.figure(figsize=(18, 12))

# Plot 1: Training Accuracy Comparison
plt.subplot(2, 3, 1)
plt.plot(history_sigmoid.history['accuracy'], label='Sigmoid - Train', linewidth=2)
plt.plot(history_sigmoid.history['val_accuracy'], label='Sigmoid - Val', linewidth=2, linestyle='--')
plt.plot(history_relu.history['accuracy'], label='ReLU - Train', linewidth=2)
plt.plot(history_relu.history['val_accuracy'], label='ReLU - Val', linewidth=2, linestyle='--')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.title('Training Progress: Sigmoid vs ReLU')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot 2: Loss Comparison
plt.subplot(2, 3, 2)
plt.plot(history_sigmoid.history['loss'], label='Sigmoid - Train', linewidth=2)
plt.plot(history_sigmoid.history['val_loss'], label='Sigmoid - Val', linewidth=2, linestyle='--')
plt.plot(history_relu.history['loss'], label='ReLU - Train', linewidth=2)
plt.plot(history_relu.history['val_loss'], label='ReLU - Val', linewidth=2, linestyle='--')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Loss: Sigmoid vs ReLU')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot 3: Gradient Magnitudes Over Layers (Sigmoid)
plt.subplot(2, 3, 3)
if grad_monitor_sigmoid.gradient_history:
    layers_to_plot = ['hidden_1', 'hidden_4', 'hidden_8']
    for layer_name in layers_to_plot:
        if layer_name in grad_monitor_sigmoid.gradient_history:
            data = grad_monitor_sigmoid.gradient_history[layer_name]
            epochs = grad_monitor_sigmoid.epoch_history[:len(data)]
            plt.plot(epochs, data, label=f'{layer_name}', marker='o', linewidth=2)

    plt.xlabel('Epoch')
    plt.ylabel('Gradient Norm')
    plt.title('Sigmoid: Gradient Magnitudes by Layer')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.yscale('log')
else:
    plt.text(0.5, 0.5, 'No gradient data collected', ha='center', va='center')

# Plot 4: Gradient Magnitudes Over Layers (ReLU)
plt.subplot(2, 3, 4)
if grad_monitor_relu.gradient_history:
    layers_to_plot = ['hidden_1', 'hidden_4', 'hidden_8']
    for layer_name in layers_to_plot:
        if layer_name in grad_monitor_relu.gradient_history:
            data = grad_monitor_relu.gradient_history[layer_name]
            epochs = grad_monitor_relu.epoch_history[:len(data)]
            plt.plot(epochs, data, label=f'{layer_name}', marker='o', linewidth=2)

    plt.xlabel('Epoch')
    plt.ylabel('Gradient Norm')
    plt.title('ReLU: Gradient Magnitudes by Layer')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.yscale('log')
else:
    plt.text(0.5, 0.5, 'No gradient data collected', ha='center', va='center')

# Plot 5: First vs Last Layer Comparison (Sigmoid)
plt.subplot(2, 3, 5)
if grad_monitor_sigmoid.gradient_history:
    first_layer = 'hidden_1'
    last_layer = f'hidden_{len([k for k in grad_monitor_sigmoid.gradient_history.keys()])}'

    if first_layer in grad_monitor_sigmoid.gradient_history and last_layer in grad_monitor_sigmoid.gradient_history:
        first_data = grad_monitor_sigmoid.gradient_history[first_layer]
        last_data = grad_monitor_sigmoid.gradient_history[last_layer]
        epochs = grad_monitor_sigmoid.epoch_history[:len(first_data)]

        plt.plot(epochs, first_data, label=f'{first_layer} (early)', linewidth=2, marker='s')
        plt.plot(epochs, last_data, label=f'{last_layer} (late)', linewidth=2, marker='^')
        plt.xlabel('Epoch')
        plt.ylabel('Gradient Norm')
        plt.title('Sigmoid: Early vs Late Layer Gradients')
        plt.legend()
        plt.grid(True, alpha=0.3)
        plt.yscale('log')
    else:
        plt.text(0.5, 0.5, 'Layer data incomplete', ha='center', va='center')
else:
    plt.text(0.5, 0.5, 'No gradient data', ha='center', va='center')

# Plot 6: Extremely Deep Network Performance
plt.subplot(2, 3, 6)
plt.plot(history_deep.history['accuracy'], label='Train Accuracy', linewidth=2)
plt.plot(history_deep.history['val_accuracy'], label='Val Accuracy', linewidth=2, linestyle='--')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.title(f'15-Layer Sigmoid Network\n(Final Val Acc: {history_deep.history["val_accuracy"][-1]:.3f})')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('vanishing_gradient_demo.png', dpi=150, bbox_inches='tight')
plt.show()

# ============================================
# 8. QUANTITATIVE ANALYSIS
# ============================================

print("\n" + "=" * 70)
print("QUANTITATIVE ANALYSIS")
print("=" * 70)

# Compare final accuracies
print(f"\nFinal Validation Accuracies:")
print(f"Sigmoid (8 layers): {history_sigmoid.history['val_accuracy'][-1]:.4f}")
print(f"ReLU (8 layers): {history_relu.history['val_accuracy'][-1]:.4f}")
print(f"Sigmoid (15 layers): {history_deep.history['val_accuracy'][-1]:.4f}")

# Check if ReLU outperformed sigmoid
if history_relu.history['val_accuracy'][-1] > history_sigmoid.history['val_accuracy'][-1]:
    print("\n✓ ReLU outperformed sigmoid, demonstrating better gradient flow")
else:
    print("\n✗ Similar performance - try deeper networks or different data")

if history_deep.history['val_accuracy'][-1] < 0.6:
    print("✓ 15-layer network shows severe vanishing gradients (poor performance)")
else:
    print("✗ 15-layer network still performing well - try even more layers")

# ============================================
# 9. SUMMARY OF FINDINGS
# ============================================

print("\n" + "=" * 70)
print("SUMMARY OF VANISHING GRADIENT OBSERVATIONS")
print("=" * 70)

print("""
Key Observations:
1. Sigmoid networks show decreasing gradient magnitudes in early layers
2. ReLU maintains stronger gradients throughout the network
3. Very deep sigmoid networks struggle to learn (accuracy near random: 0.5)
4. The gradient norm decreases exponentially with layer depth in sigmoid networks

Solutions to Vanishing Gradients:
- Use ReLU or Leaky ReLU activation
- Add Batch Normalization layers
- Use Residual connections (skip connections)
- Proper weight initialization (He initialization for ReLU)
- Use architectures like ResNet, DenseNet specifically designed for deep networks
""")
```
